```python

#monthList  = df ['month_number'].tolist()
#bathingsoap   = df ['bathingsoap'].tolist()
#faceWashSalesData   = df ['facewash'].tolist()
x  = df ['month_number'].tolist()
y   = df ['bathingsoap'].tolist()
y1  = df ['facewash'].tolist()
y2   = df ['toothpaste'].tolist()
y3  = df ['shampoo'].tolist()
f, axarr = plt.subplots(3, sharex=True)
axarr[0].plot(x, y, label = 'Bathingsoap Sales Data', color='k', marker='o', linewidth=3)
axarr[0].set_title('Sales data of  a Bathingsoap')
axarr[1].plot(x, y1, label = 'Face Wash Sales Data', color='r', marker='o', linewidth=3)
axarr[1].set_title('Sales data of  a facewash')

plt.xticks(x)
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
axarr[2].stackplot(x,y,y1,y2,y3, colors=['r','g','y','b'])
plt.show()
```
